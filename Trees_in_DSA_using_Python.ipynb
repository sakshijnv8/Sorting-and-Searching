# **Terminology in Trees**

- #  Parent Node :

    The node which is an immediate predecessor of a node is called the parent node of that node. {B} is the parent node of {D, E}.
- # Child Node:

    The node which is the immediate successor of a node is called the child node of that node. Examples: {D, E} are the child nodes of {B}.
- # Root Node:

    The topmost node of a tree or the node which does not have any parent node is called the root node. {A} is the root node of the tree. A non-empty tree must contain exactly one root node and exactly one path from the root to all other nodes of the tree.
- # Leaf Node or External Node:

    The nodes which do not have any child nodes are called leaf nodes. {I, J, K, F, G, H} are the leaf nodes of the tree.
- # Ancestor of a Node:

    Any predecessor nodes on the path of the root to that node are called Ancestors of that node. {A,B} are the ancestor nodes of the node {E}
- # Descendant:

    A node x idns a descendant of another node y if and only if y is an ancestor of x.
- # Sibling:

    Children of the same parent node are called siblings. {D,E} are called siblings.
- # Level of a node:

    The count of edges on the path from the root node to that node. The root node has level 0.
- # Internal node:

    A node with at least one child is called Internal Node.
- # Neighbour of a Node:

    Parent or child nodes of that node are called neighbors of that node.
- # Subtree:

    Any node of the tree along with its descendant.

Preorder traversal is a way to visit all nodes of a binary tree in this order:

- Root → Left → Right

Visit current node first

Traverse left subtree

Traverse right subtree

# **Pre Order Traversal using Stacks**

In real production systems:

recursion depth can overflow

iterative solutions preferred sometimes

So, we use STACK internally.

##**The profit of using a stack in DFS is that it efficiently remembers the path for backtracking, avoids recursion, and enables depth-first traversal in O(n) time.**

````python
class TreeNode:

    def __init__(self, value):

        self.value = value
        self.left = None
        self.right = None


def preorder_iterative(root):

    if root is None:
        return

    stack = [root]

    while stack:

        current = stack.pop()

        print(current.value, end=" ")    # first add root value after the pop

        # Push right first because stack works on lifo order so left add in last
        if current.right:     #if current.right not equal to none than
            stack.append(current.right)

        # Push left second
        if current.left:
            stack.append(current.left)  #this pop first because add in last so append in stack before right node

root = TreeNode("A")

root.left = TreeNode("B")
root.right = TreeNode("C")

root.left.left = TreeNode("D")
root.left.right = TreeNode("E")

root.right.left = TreeNode("F")
root.right.right = TreeNode("G")

preorder_iterative(root)
````

#**Binary Tree Preorder Traversal**

- https://leetcode.com/problems/binary-tree-preorder-traversal/

````python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right

class Solution:

    def preorderTraversal(self, root):

        result = []

        def dfs(node):

            # Base condition
            if node is None:
                return

            # Step 1: Visit root
            result.append(node.val)

            # Step 2: Traverse left subtree
            dfs(node.left)

            # Step 3: Traverse right subtree
            dfs(node.right)

        dfs(root)

        return result
````

## Post Order Traversal using Stack (Implementation):

# Intuition

In Postorder Traversal we need:

```text id="ssg9r2"
Left → Right → Root
```

But stack naturally processes in reverse order because it follows:

```text id="sd2k1l"
LIFO (Last In First Out)
```

So instead of directly trying to achieve Postorder, we do something easier first.

We create this order:

```text id="o8ic17"
Root → Right → Left
```

using `stack1`.

Then we store nodes into `stack2`.

When we later pop from `stack2`, the order automatically reverses into:

```text id="3oz3ql"
Left → Right → Root
```

which is exactly Postorder Traversal.

---

# Core Idea

```text id="3h5v6w"
First Stack:
Create reverse postorder

Second Stack:
Reverse it again to get actual postorder
```

---

# Visualization

Desired:

```text id="ivylbp"
Left → Right → Root
```

Generated first:

```text id="3zh9ma"
Root → Right → Left
```

Reverse it:

```text id="x6d1e4"
Left → Right → Root
```

---

# Brief Algorithm

## Step 1

If tree is empty:

```python id="y4b1j0"
return []
```

---

## Step 2

Create:

* `stack1` → for processing nodes
* `stack2` → for reversing order

Push root into `stack1`.

---

## Step 3

While `stack1` is not empty:

### a)

Pop node from `stack1`.

### b)

Push node into `stack2`.

### c)

Push left child into `stack1`.

### d)

Push right child into `stack1`.

---

# Why Left First Then Right?

Because stack is LIFO.

Right child gets processed first.

So generated order becomes:

```text id="l1ol9d"
Root → Right → Left
```

---

## Step 4

Pop all nodes from `stack2`.

Add them into result.

This reverses order into:

```text id="2gthv8"
Left → Right → Root
```

---

# Final Complexity

## Time Complexity

```text id="0v11a2"
O(N)
```

Each node is pushed and popped constant number of times.

---

## Space Complexity

```text id="cqptlf"
O(N)
```

because two stacks are used.

---

# One-Line Memory Trick

```text id="h5sxf9"
Root Right Left
↓ reverse
Left Right Root
```

That is the whole intuition behind iterative Postorder using 2 stacks.


````python
# stack post order implementation:
class Solution:

    def postorderTraversal(self, root):

        if root is None:
            return []

        stack1 = [root]
        stack2 = []

        result = []

        while stack1:

            node = stack1.pop()

            stack2.append(node)

            if node.left:
                stack1.append(node.left)

            if node.right:
                stack1.append(node.right)

        while stack2:

            result.append(stack2.pop().value)

        return result

````

````python
root = TreeNode(1)

root.left = TreeNode(2)
root.right = TreeNode(3)

root.left.left = TreeNode(4)
root.left.right = TreeNode(5)

root.right.left = TreeNode(6)
root.right.right = TreeNode(7)


# Execute

obj = Solution()

answer = obj.postorderTraversal(root)

print(answer)
````

````python
#stack implementation

class Solution:

    def inorderTraversal(self, root):

        stack = []

        result = []

        current = root

        while current or stack:

            # Go left completely
            while current:

                stack.append(current)

                current = current.left

            # Process node
            current = stack.pop()

            result.append(current.value)

            # Move right
            current = current.right

        return result


# create Binary Tree
root = TreeNode(1)

root.left = TreeNode(2)
root.right = TreeNode(3)

root.left.left = TreeNode(4)
root.left.right = TreeNode(5)

root.right.left = TreeNode(6)
root.right.right = TreeNode(7)


# Execute

obj = Solution()

answer = obj.inorderTraversal(root)

print(answer)
````

# **Level Order Traversal**
### Leet Code 102. Binary Tree Level Order Traversal
- https://leetcode.com/problems/binary-tree-level-order-traversal/

## Algorithm:

Step 1: If tree empty: return []

Step 2: Create queue. Add root node -> queue = [root]

Step 3: Run loop until queue becomes empty.
- while queue:

Step 4: Find current level size. level_size = len(queue)

Step 5: Process exactly those many nodes.
- for i in range(level_size):

Step 6: For every node:

- 1. Remove node from queue
- 2. Store node value
- 3. Add left child
- 4. Add right child

Step 7: Store completed level in answer.

````python
from collections import deque

class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None


class Solution:
    def levelOrder(self, root):

        result = []

        # Edge case
        if not root:
            return result

        # Queue for BFS
        queue = deque([root])

        # Traverse until queue becomes empty
        while queue:

            # Number of nodes in current level
            level_size = len(queue)

            # Store nodes of current level
            current_level = []

            # Process all nodes of same level
            for i in range(level_size):

                # Remove node from front
                node = queue.popleft()

                # Store node value
                current_level.append(node.val)

                # Add left child
                if node.left:
                    queue.append(node.left)

                # Add right child
                if node.right:
                    queue.append(node.right)

            # Add current level to result
            result.append(current_level)

        return result


# result = [[3], [9,20], [15,7]]
# result = [3,9,20]

# create the tree

root = TreeNode(3)

root.left = TreeNode(9)

root.right = TreeNode(20)

root.right.left = TreeNode(15)

root.right.right = TreeNode(7)


# now let's call the function to execute the code

obj = Solution()

answer = obj.levelOrder(root)

print(answer)
````

#**LeetCode 103: Binary Tree Zigzag Level Order Traversal**

- https://leetcode.com/problems/binary-tree-zigzag-level-order-traversal/

The only difference from normal level-order traversal is:

One level goes left → right

Next level goes right → left

Then left → right again

Like a zigzag.

Level 0 → left to right

Level 1 → right to left

Level 2 → left to right

Level 3 → right to left


### Algorithm:

1. Perform normal BFS
2. Store level nodes
3. Reverse alternate levels

````python
from collections import deque


class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None


class Solution:
    def zigzagLevelOrder(self, root):

        if not root:
            return []

        ans = []

        queue = deque([root])

        while queue:

            level = []

            level_size = len(queue)

            # Process current level
            for i in range(level_size):

                node = queue.popleft()

                level.append(node.val)

                if node.left:
                    queue.append(node.left)

                if node.right:
                    queue.append(node.right)

            # Reverse alternate levels
            if len(level) > 0:

                if len(ans) % 2 == 1:
                    level.reverse()

                ans.append(level)

        return ans


# CREATE TREE

root = TreeNode(3)

root.left = TreeNode(9)

root.right = TreeNode(20)

root.right.left = TreeNode(15)

root.right.right = TreeNode(7)



obj = Solution()

print(obj.zigzagLevelOrder(root))
````

#**LeetCode 107. Binary Tree Level Order Traversal II**
- https://leetcode.com/problems/binary-tree-level-order-traversal-ii/

````python
from collections import deque


class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None


class Solution:

    def levelOrderBottom(self, root):

        # Final answer
        result = []

        # Edge case
        if not root:
            return result

        # Queue for BFS
        queue = deque([root])

        # BFS traversal
        while queue:

            # Nodes in current level
            level_size = len(queue)

            current_level = []

            # Process current level
            for i in range(level_size):

                node = queue.popleft()

                current_level.append(node.val)

                # Add left child
                if node.left:
                    queue.append(node.left)

                # Add right child
                if node.right:
                    queue.append(node.right)

            # Store current level
            result.append(current_level)

        # Reverse final answer
        result.reverse()

        return result
````

# **637. Average of Levels in Binary Tree**

- https://leetcode.com/problems/average-of-levels-in-binary-tree/


This is STILL: BFS + Queue
because wee still need level-by-level traversal

The ONLY difference is instead of storing all level nodes permanently:

We calculate sum and average

For every level:

1. Find total sum
2. Find number of nodes
3. Average = sum / count


## **Algorithm**
Step 1: If tree empty:
return []

Step 2: Create queue and add root.

Step 3: Run BFS until queue empty.

Step 4: For each level:
- Get: level_size = len(queue)
- This tells: How many nodes belong to current level

Step 5: Initialize: level_sum = 0

Step 6: Process all nodes of current level.
- Add node values into: level_sum

Step 7: Compute average:
Average = Level Sum/ Number of Nodes

Step 8: Store average in result.


## Time Complexity
Each node visited once: O(n)

## Space Complexity
Queue stores nodes: O(n)

````python
from collections import deque


class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None


class Solution:

    def averageOfLevels(self, root):

        # Final answer
        result = []

        # Edge case
        if not root:
            return result

        # Queue for BFS
        queue = deque([root])

        # BFS traversal
        while queue:

            # Number of nodes in current level
            level_size = len(queue)

            # Sum of current level
            level_sum = 0

            # Process current level
            for i in range(level_size):

                node = queue.popleft()

                # Add node value into sum
                level_sum += node.val

                # Add left child
                if node.left:
                    queue.append(node.left)

                # Add right child
                if node.right:
                    queue.append(node.right)

            # Calculate average
            average = level_sum / level_size

            # Store answer
            result.append(average)
            # result.append(level_sum)

        return result
````

````python
# max of each level


from collections import deque


class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None


class Solution:

    def averageOfLevels(self, root):

        # Final answer
        result = []

        # Edge case
        if not root:
            return result

        # Queue for BFS
        queue = deque([root])

        # BFS traversal
        while queue:

            # Number of nodes in current level
            level_size = len(queue)
            current_level = []

            # Process current level
            for i in range(level_size):

                node = queue.popleft()
                current_level.append(node.val)

                # Add left child
                if node.left:
                    queue.append(node.left)

                # Add right child
                if node.right:
                    queue.append(node.right)


            level_max = max(current_level)
            result.append(level_max)

        return result

# CREATE TREE

root = TreeNode(3)

root.left = TreeNode(9)

root.right = TreeNode(20)

root.right.left = TreeNode(15)

root.right.right = TreeNode(7)


# RUN

obj = Solution()

print(obj.averageOfLevels(root))
````

````python
from collections import deque

class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None


class Solution:
    def rightview(self, root):

        result = []

        # Edge case
        if not root:
            return result

        # Queue for BFS
        queue = deque([root])

        # Traverse until queue becomes empty
        while queue:

            # Number of nodes in current level
            level_size = len(queue)

            # Store nodes of current level
            current_level = []

            # Process all nodes of same level
            for i in range(level_size):

                # Remove node from front
                node = queue.popleft()

                # Store node value
                current_level.append(node.val)

                # Add left child
                if node.left:
                    queue.append(node.left)

                # Add right child
                if node.right:
                    queue.append(node.right)

            # Add current level to result
            result.append(current_level[-1])

        return result


# result = [[3], [9,20], [15,7]]
# result = [3,9,20]

# create the tree

root = TreeNode(3)

root.left = TreeNode(9)

root.right = TreeNode(20)

root.right.left = TreeNode(15)

root.right.right = TreeNode(7)


# now let's call the function to execute the code

obj = Solution()

answer = obj.rightview(root)

print(answer)
````

# **101. Symmetric Tree**
- https://leetcode.com/problems/symmetric-tree/


## Mirror Condition

For two nodes to be mirror:

### Condition 1: Values must be same.

###Condition 2: Left child of first tree must match the Right child of second tree

###Condition 3: Right child of first tree must match the Left child of second tree

---
# LOGIC

Left.Left=Right.Right

AND

Left.Right=Right.Left

This is the WHOLE problem.

---
## Algorithm

For two nodes left and right:

## Step 1: If both are None: return True

## Step 2: If one is None: return False

## Step 3: If values different: return False

## Step 4: Recursively compare:

left.left  ↔ right.right

AND

left.right ↔ right.left


Screenshot 2026-05-13 at 11.16.35 PM.png

````python
# recursion logic

class Solution:

    def isSymmetric(self, root):

        def isMirror(left, right):

            # Both empty
            if left is None and right is None:
                return True

            # One empty
            if left is None or right is None:
                return False

            # Values different
            if left.val != right.val:
                return False

            # Mirror comparison
            return (
                isMirror(left.left, right.right)
                and
                isMirror(left.right, right.left)
            )

        return isMirror(root.left, root.right)
````

````python
# queue implementation to check symmetric tree

from collections import deque

class Solution:

    def isSymmetric(self, root):

        queue = deque()

        queue.append((root.left, root.right))

        while queue:

            left, right = queue.popleft()

            # Both None
            if left is None and right is None:
                continue

            # One None
            if left is None or right is None:
                return False

            # Values different
            if left.val != right.val:
                return False

            # Push mirror pairs
            queue.append((left.left, right.right))
            queue.append((left.right, right.left))

        return True
````

#**110. Balanced Binary Tree**

- https://leetcode.com/problems/balanced-binary-tree/

A tree is balanced if:

For EVERY node -

Difference between left subtree height and right subtree height is NOT more than 1.

Balanced trees are CRITICAL because the operations become: O(logn)

## At EVERY node:
calculate left height

calculate right height

compare difference

## BRUTE FORCE IDEA : (Not optimal)

For every node:

calculate left height

calculate right height

recursively check children

###Problem in this approach

Height calculation repeated many times

Time complexity becomes: O(n2)


---
# ALGORITHM

## Steps:
- For every node:

##Step 1: Find left subtree height.

## Step 2: Find right subtree height.

## Step 3: If difference > 1: return -1

## Step 4 Else return:  1+max(leftHeight,rightHeight)

````python
class Solution:

    def isBalanced(self, root):

        def height(node):

            # Empty tree height
            if node is None:
                return 0

            # Left subtree height
            left = height(node.left)

            # If already unbalanced
            if left == -1:
                return -1

            # Right subtree height
            right = height(node.right)

            # If already unbalanced
            if right == -1:
                return -1

            # Check balance condition
            if abs(left - right) > 1:
                return -1

            # Return current height
            return 1 + max(left, right)

        return height(root) != -1
````

# Algorithm — Insert into a Binary Search Tree

## Step 1: If tree is empty: return new node

## Step 2: Compare value with current node.

## Step 3: If: value < root.val -> move LEFT subtree.

## Step 4: Else: value > root.val -> move RIGHT subtree.

## Step 5: Repeat recursively until empty position found.

## Step 6: Insert new node there.


````python
# solution 1: to insert a new node into an existing BST

class TreeNode:

    def __init__(self, val):

        self.val = val
        self.left = None
        self.right = None


# Iterative BST Insert

class Solution:

    def insertIntoBST(self, root, val):

        # If tree is empty
        if root is None:

            return TreeNode(val)

        newNode = TreeNode(val)

        curr = root

        while curr is not None:

            # Go LEFT
            if val < curr.val:

                # Move left
                if curr.left is not None:

                    curr = curr.left

                # Insert here
                else:

                    curr.left = newNode

                    break

            # Go RIGHT
            else:

                # Move right
                if curr.right is not None:

                    curr = curr.right

                # Insert here
                else:

                    curr.right = newNode

                    break

        return root


# Inorder Traversal
def inorder(root):

    if root is None:
        return

    inorder(root.left)

    print(root.val, end=" ")

    inorder(root.right)


root = TreeNode(50)

root.left = TreeNode(30)
root.right = TreeNode(70)

root.left.left = TreeNode(20)
root.left.right = TreeNode(40)


# Before Insertion

print("Before Insertion:")

inorder(root)

print()

# Insert New Value

obj = Solution()

root = obj.insertIntoBST(root, 60)


# After Insertion

print("\nAfter Inserting 60:")

inorder(root)
````

# 700. Search in a Binary Search Tree
- https://leetcode.com/problems/search-in-a-binary-search-tree/description/


At every step: eliminate HALF tree. That is why BST becomes efficient.

## Search Algorithm

Suppose searching value x.

Step 1: If root is None: return None
Step 2: If: root.value == x. Found. Return node.

Step 3: If: x < root.value. Search LEFT subtree.

Step 4: Else: Search RIGHT subtree


````python
# Recursive Python Code
class TreeNode:

    def __init__(self, value):

        self.value = value

        self.left = None
        self.right = None


def search(root, target):

    # Base case
    if root is None:

        return None

    # Found target
    if root.value == target:

        return root

    # Search LEFT
    if target < root.value:

        return search(root.left, target)

    # Search RIGHT
    else:

        return search(root.right, target)


# Create BST

root = TreeNode(50)

root.left = TreeNode(30)
root.right = TreeNode(70)

root.left.left = TreeNode(20)
root.left.right = TreeNode(40)

root.right.left = TreeNode(60)
root.right.right = TreeNode(80)



result = search(root, 60)

if result:
    print("Found:", result.value)

else:
    print("Not Found")
````

````python
# iterative solution (alternate)- preferred for large trees

def search_iterative(root, target):

    current = root

    while current:

        # Found
        if current.value == target:

            return current

        # Go LEFT
        elif target < current.value:

            current = current.left

        # Go RIGHT
        else:

            current = current.right

    return None



# Create BST

root = TreeNode(50)

root.left = TreeNode(30)
root.right = TreeNode(70)

root.left.left = TreeNode(20)
root.left.right = TreeNode(40)

root.right.left = TreeNode(60)
root.right.right = TreeNode(80)



result = search_iterative(root, 60)

if result:
    print("Found:", result.value)

else:
    print("Not Found")
````

````python
# delete the given node from BST


class Solution:

    def deleteNode(self, root, key):

        # Tree empty
        if root is None:

            return None

        # Go LEFT
        if key < root.val:

            root.left = self.deleteNode(root.left, key)

        # Go RIGHT
        elif key > root.val:

            root.right = self.deleteNode(root.right, key)

        # Node Found
        else:

            # if root.left is None and root.right is None:
              # return None

            # CASE 1 : No left child
            if root.left is None:

                return root.right

            # CASE 2 : No right child
            elif root.right is None:

                return root.left

            # CASE 3 : Two children
            else:

                # Find inorder successor - means jo node delete krni hai uske just baad kon aega (inorder traversal me)
                # it'll always be the minimum node in the right subtree
                successor = self.findMin(root.right)

                # Replace value
                root.val = successor.val

                # Delete successor
                root.right = self.deleteNode(
                    root.right,
                    successor.val
                )

        return root


    # Find minimum node
    def findMin(self, node):

        while node.left:

            node = node.left

        return node
````

# **Validate Binary Search Tree**

You are given a binary tree.

You need to check:

Is this tree a VALID BST?
MOST IMPORTANT BST RULE

For every node:

Left Subtree<Root<Right Subtree

AND this rule must hold for:

EVERY node

not just immediate children.

2. VERY IMPORTANT INTUITION

Most beginners make this mistake:

They only check:

left child < root < right child

That is NOT enough.

Example of INVALID BST
          10
         /  \
        5    15
            /  \
           6    20

Looks valid locally.

But:

6 < 10

and 6 exists in RIGHT subtree of 10.

So tree is INVALID.

MOST IMPORTANT UNDERSTANDING

Every node must satisfy:

range constraints

Example:

left subtree nodes must remain smaller
right subtree nodes must remain larger

even deep inside tree.

3. BEST INTUITION (Range Method)

Think:

Each node receives:

minimum allowed value
maximum allowed value
Example

For root 10:

(-∞ , +∞)
Left Child 5

Allowed range:

(-∞ , 10)
Right Child 15

Allowed range:

(10 , +∞)
Node 6

Since 6 is inside right subtree of 10:

Allowed range becomes:

(10 , 15)

But:

6 < 10

Invalid BST.

4. Algorithm
STEP 1

Start with:

(-infinity, +infinity)
STEP 2

For every node:

value must lie within range
STEP 3

Go LEFT:

update max boundary
STEP 4

Go RIGHT:

update min boundary
STEP 5

If any node violates range:

````python
class Solution:

    def isValidBST(self, root):

        def validate(node, low, high):

            # Empty tree valid
            if node is None:
                return True

            # BST violation
            if not (low < node.val < high):
                return False

            # Validate LEFT subtree
            left = validate(
                node.left,
                low,
                node.val
            )

            # Validate RIGHT subtree
            right = validate(
                node.right,
                node.val,
                high
            )

            return left and right

        return validate(
            root,
            float("-inf"),
            float("inf")
        )
````

````python
from collections import deque


# -----------------------------------
# File/Folder Node
# -----------------------------------

class FileNode:

    def __init__(self, name, is_file=False):

        self.name = name
        self.is_file = is_file

        self.children = []


# -----------------------------------
# Smart File Explorer
# -----------------------------------

class FileExplorer:

    def __init__(self):

        self.root = FileNode("Root")

        # Stack for undo delete
        self.undo_stack = []


    # -----------------------------------
    # Add child node
    # -----------------------------------

    def add(self, parent, child):

        parent.children.append(child)


    # -----------------------------------
    # Print File Hierarchy
    # -----------------------------------

    def print_structure(self, node=None, level=0):

        if node is None:
            node = self.root

        print(" " * level + node.name)

        for child in node.children:

            self.print_structure(child, level + 4)


    # -----------------------------------
    # DFS Search
    # -----------------------------------

    def dfs_search(self, node, target):

        if node.name == target:

            return True

        for child in node.children:

            found = self.dfs_search(child, target)

            if found:
                return True

        return False


    # -----------------------------------
    # BFS Search
    # -----------------------------------

    def bfs_search(self, target):

        queue = deque([self.root])

        while queue:

            current = queue.popleft()

            if current.name == target:

                return True

            for child in current.children:

                queue.append(child)

        return False


    # -----------------------------------
    # Delete Node
    # -----------------------------------

    def delete(self, parent, target):

        for child in parent.children:

            if child.name == target:

                # Push to undo stack
                self.undo_stack.append((parent, child))

                parent.children.remove(child)

                print(f"{target} deleted")

                return

            self.delete(child, target)


    # -----------------------------------
    # Undo Delete
    # -----------------------------------

    def undo_delete(self):

        if not self.undo_stack:

            print("Nothing to undo")
            return

        parent, node = self.undo_stack.pop()

        parent.children.append(node)

        print(f"{node.name} restored")
````

````python
# -----------------------------------
# Create Explorer
# -----------------------------------

explorer = FileExplorer()


# -----------------------------------
# Create Folder Structure
# -----------------------------------

documents = FileNode("Documents")
photos = FileNode("Photos")
projects = FileNode("Projects")

resume = FileNode("Resume.pdf", True)
notes = FileNode("Notes.txt", True)

trip = FileNode("Trip.jpg", True)

python_folder = FileNode("Python")


# -----------------------------------
# Build Tree
# -----------------------------------

explorer.add(explorer.root, documents)
explorer.add(explorer.root, photos)
explorer.add(explorer.root, projects)

explorer.add(documents, resume)
explorer.add(documents, notes)

explorer.add(photos, trip)

explorer.add(projects, python_folder)


# -----------------------------------
# Print Structure
# -----------------------------------

print("\nFILE STRUCTURE:\n")

explorer.print_structure()


# -----------------------------------
# DFS Search
# -----------------------------------

print("\nDFS Search:\n")

print(explorer.dfs_search(explorer.root, "Resume.pdf"))


# -----------------------------------
# BFS Search
# -----------------------------------

print("\nBFS Search:\n")

print(explorer.bfs_search("Trip.jpg"))


# -----------------------------------
# Delete Folder
# -----------------------------------

print("\nDELETE:\n")

explorer.delete(explorer.root, "Photos")

explorer.print_structure()


# -----------------------------------
# Undo Delete
# -----------------------------------

print("\nUNDO DELETE:\n")

explorer.undo_delete()

explorer.print_structure()
````
